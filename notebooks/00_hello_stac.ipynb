```python
from fcl.config import load_config
from fcl.stac import search_s1, search_s2

cfg = load_config("configs/aoi.yaml")
cfg["bbox"]
```

```python
s2 = search_s2(cfg, start="2024-06-01", end="2024-08-31", max_cloud=20)
print(len(s2), "Sentinel-2 L2A scenes (summer 2024, <20% cloud)")
for it in list(s2)[:5]:
    print(it.id, it.datetime.date(), round(it.properties["eo:cloud_cover"], 1))
```

```python
s1 = search_s1(cfg, start="2024-06-01", end="2024-08-31")
print(len(s1), "Sentinel-1 RTC scenes (summer 2024)")
for it in list(s1)[:5]:
    print(it.id, it.datetime.date(), it.properties.get("sat:orbit_state"))
```

If both counts are > 0, step 1 is done. ✅
